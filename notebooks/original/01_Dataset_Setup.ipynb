```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

# Our folders
img_folder = '/content/drive/MyDrive/Brain_Tumor_Project/brisc2025/segmentation_task/train/images'
mask_folder = '/content/drive/MyDrive/Brain_Tumor_Project/brisc2025/segmentation_task/train/masks'

# Get the lists of files and sort them so they match up perfectly
all_images = sorted(os.listdir(img_folder))
all_masks = sorted(os.listdir(mask_folder))

# Grab the name of the very first picture (number 0 in computer language)
first_img_name = all_images[0]
first_mask_name = all_masks[0]

# Connect the folder path to the picture name
img_path = os.path.join(img_folder, first_img_name)
mask_path = os.path.join(mask_folder, first_mask_name)

# Read the pictures
img = mpimg.imread(img_path)
mask = mpimg.imread(mask_path)

# Draw them side-by-side!
fig, (box1, box2) = plt.subplots(1, 2, figsize=(10, 5))

box1.imshow(img, cmap='gray')
box1.set_title("Original MRI Scan")

box2.imshow(mask, cmap='gray')
box2.set_title("Tumor Location (Mask)")

plt.show()
```
